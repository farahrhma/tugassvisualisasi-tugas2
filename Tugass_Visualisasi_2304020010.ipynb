# Bar Plot Jumlah Berdasarkan SKill

```python
mapping = {
    'Weak': 1,
    'Average': 2,
    'Strong': 3
}
df_skill = df[['Python', 'Java', 'SQL']].replace(mapping)
rata_skill = df_skill.mean()
plt.bar(rata_skill.index, rata_skill.values)

plt.grid(axis='y', alpha=0.5)
for i in range(len(rata_skill)):
    plt.text(i, rata_skill.values[i],
             round(rata_skill.values[i],2),
             ha='center', va='bottom')

plt.xlabel('Jenis Skill')
plt.ylabel('Rata-rata Level Skill (1=Weak, 3=Strong)')
plt.title('Rata-rata Skill Programming Mahasiswa')

plt.show()
```

# Bar Plot dengan Transparansi & Label Angka

```python
mapping = {'Weak':1, 'Average':2, 'Strong':3}
df_skill = df[['Python','Java','SQL']].replace(mapping)

rata_skill = df_skill.mean()
kategori = rata_skill.index
data_bar = rata_skill.values

plt.figure(figsize=(8,5))

plt.barh(kategori, data_bar,
         color='red',
         alpha=0.25)

plt.grid(linestyle='-',
         linewidth=2,
         axis='x',
         alpha=0.5)

for i in range(len(data_bar)):
    plt.text(data_bar[i], i,
             round(data_bar[i],2),
             va='center')

plt.ylabel('Jenis Skill')
plt.xlabel('Rata-rata Level Skill')
plt.title('Rata-rata Skill Programming Mahasiswa')

plt.show()
```

# Stacked Bar Plot (Skill berdasarkan Gender)

```python
cross = pd.crosstab(df['Python'], df['Gender'])

data1 = cross.iloc[:,0]
data2 = cross.iloc[:,1]
kategori = cross.index

plt.bar(kategori, data1, label=cross.columns[0])
plt.bar(kategori, data2, label=cross.columns[1], bottom=data1)

plt.legend()

plt.grid(linestyle='-',
         linewidth=2,
         axis='y',
         alpha=0.5)

plt.xlabel('Level Skill Python')
plt.ylabel('Jumlah')
plt.title('Skill Python Berdasarkan Gender')

plt.show()
```

```python
cross = pd.crosstab(df['SQL'], df['Gender'])

data1 = cross.iloc[:,0]
data2 = cross.iloc[:,1]
kategori = cross.index

plt.bar(kategori, data1, label=cross.columns[0])
plt.bar(kategori, data2, label=cross.columns[1], bottom=data1)

plt.legend()

plt.grid(linestyle='-',
         linewidth=2,
         axis='y',
         alpha=0.5)

plt.xlabel('Level Skill SQL')
plt.ylabel('Jumlah')
plt.title('Skill Python Berdasarkan Gender')

plt.show()
```

```python
cross = pd.crosstab(df['Java'], df['Gender'])

data1 = cross.iloc[:,0]
data2 = cross.iloc[:,1]
kategori = cross.index

plt.bar(kategori, data1, label=cross.columns[0])
plt.bar(kategori, data2, label=cross.columns[1], bottom=data1)

plt.legend()

plt.grid(linestyle='-',
         linewidth=2,
         axis='y',
         alpha=0.5)

plt.xlabel('Level Skill Java')
plt.ylabel('Jumlah')
plt.title('Skill Python Berdasarkan Gender')

plt.show()
```

# Grouped Bar Plot (skill programming)

```python
skill_map = {
    'Weak':1,
    'Average':2,
    'Strong':3
}

df['Python_num'] = df['Python'].str.strip().map(skill_map)
df['Java_num'] = df['Java'].str.strip().map(skill_map)
df['SQL_num'] = df['SQL'].str.strip().map(skill_map)
```

```python
data1 = [df['Python_num'].mean(), df['Java_num'].mean(), df['SQL_num'].mean()]
data2 = [df['Python_num'].max(), df['Java_num'].max(), df['SQL_num'].max()]

kategori = ['Python','Java','SQL']

x = np.arange(len(kategori))
width = 0.35

plt.bar(x-width/2, data1, width, label='Rata-rata Skill')
plt.bar(x+width/2, data2, width, label='Skill Maksimum')

plt.xticks(x, kategori)

plt.grid(linestyle='--',
         linewidth=1,
         axis='y',
         alpha=0.75)

plt.xlabel('Bahasa Pemrograman')
plt.ylabel('Level Skill')
plt.title('Perbandingan Skill Programming Mahasiswa')

plt.legend()

plt.show()
```

# Pie Chart tanpa explode (skill)

```python
mapping = {'Weak':1, 'Average':2, 'Strong':3}
df_skill = df[['Python','Java','SQL']].replace(mapping)

rata_skill = df_skill.mean()

kategori = rata_skill.index
data_pie = rata_skill.values

warna = ['red','blue','green']

plt.figure(figsize=(6,6))

plt.pie(data_pie,
        labels=kategori,
        colors=warna,
        autopct='%1.1f%%',
        startangle=90)

plt.legend()
plt.title('Persentase Skill Mahasiswa')

plt.show()
```

# Exploded Pie Chart (Skill)

```python
mapping = {'Weak':1, 'Average':2, 'Strong':3}
df_skill = df[['Python','Java','SQL']].replace(mapping)

rata_skill = df_skill.mean()

kategori = rata_skill.index
data_bar = rata_skill.values
warna = ['red','blue','green','orange','purple']
explode_var = [0.1] + [0]*(len(data_bar)-1)

plt.pie(data_bar,
        labels=kategori,
        colors=warna[:len(data_bar)],
        explode=explode_var,
        autopct='%1.1f%%',
        startangle=90)

plt.legend()
plt.title('Persentase Skill Mahasiswa')

plt.show()
```
